# Pollinations.AI Image Generation Demo

## English 🇬🇧
This notebook shows how to generate AI images for your projects using Pollinations.AI.

- No signup or API key needed
- Works in any language (see Chinese example below)

## 中文 🇨🇳
本教程演示如何用 Pollinations.AI 免费生成 AI 图片。

- 无需注册或 API 密钥
- 支持多语言（见下方中文示例）


```python
import requests
from IPython.display import Image, display

# English prompt
prompt_en = 'A robot teaching kids about AI'
url_en = f'https://image.pollinations.ai/prompt/{prompt_en.replace(' ', '%20')}'
response_en = requests.get(url_en)
with open('robot_ai_teacher_en.jpg', 'wb') as f:
    f.write(response_en.content)
display(Image('robot_ai_teacher_en.jpg'))
```

```python
# 中文提示词
prompt_cn = '一个机器人正在给孩子们讲解人工智能'
url_cn = f'https://image.pollinations.ai/prompt/{prompt_cn}'
response_cn = requests.get(url_cn)
with open('robot_ai_teacher_cn.jpg', 'wb') as f:
    f.write(response_cn.content)
display(Image('robot_ai_teacher_cn.jpg'))
```
